```python
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph,START,END
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
```

```python
llm = ChatOpenAI()
```

```python
class BlogLLM(TypedDict):
    title : str
    outline : str
    blog_content : str
```

```python
def gen_outline(BlogLLM) -> BlogLLM:

    topic = BlogLLM['title']

    prompt = f"Generate an outline for this tile -> {topic}"

    response = llm.invoke(prompt)

    BlogLLM['outline'] = response.content

    return BlogLLM
```

```python
def gen_blog(BlogLLM) -> BlogLLM:

    topic = BlogLLM['title']
    outline = BlogLLM['outline']

    prompt = f"Generate a blog on the topic -> {topic} with following outline {outline} in 50 words."

    response = llm.invoke(prompt)

    BlogLLM['blog_content'] = response.content

    return BlogLLM 
```

```python
graph = StateGraph(BlogLLM)
```

```python
#  add a node
graph.add_node('gen_outline',gen_outline)
graph.add_node('gen_blog',gen_blog)

#  add edges

graph.add_edge(START,'gen_outline')
graph.add_edge('gen_outline','gen_blog')
graph.add_edge('gen_blog',END)

# Compile Graph

workflow = graph.compile()
```

```python
initial_state = {'title':'Advantages of Being Delusional'}

final_state = workflow.invoke(initial_state)

print(final_state)
```
