```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BMIState(TypedDict):

    weight_kg : float
    height_m : float
    bmi : float
```

```python
# create a python function for graph to calculate BMI

def calculate_bmi(BMIState) -> BMIState:
    weight = BMIState['weight_kg']
    height = BMIState['height_m']
    bmi = weight / (height**2)

    BMIState['bmi'] = round(bmi,2)

    #  this is known as partial update
    return BMIState 
```

```python
#  create graph
graph = StateGraph(BMIState)

#  add nodes

graph.add_node('calculate_bmi',calculate_bmi)

#  add edges
graph.add_edge(START,'calculate_bmi')
graph.add_edge('calculate_bmi',END)

#  compile the graph

workflow = graph.compile()
```

```python
# execture the graph
initial_state = {'weight_kg' : 80 , 'height_m' : 1.73}

final_state = workflow.invoke(initial_state)

print(final_state)
```
