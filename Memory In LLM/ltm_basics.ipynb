# About

1. I'll learn about creation of memory stores and use of in-memory stores.

```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
namespace = ('user','u1')
```

```python
namespace2 = ('user','u2')
```

# Creating Memories

```python
store.put(namespace,"1",{"data": "User Likes Pizza"})
store.put(namespace,"2" , {"data" : "User likes Pasta!"})
```

```python
store.put(namespace2,"1",{"data": "User prefers I phone"})
store.put(namespace2,"2" , {"data" : "User preferred coding language is Python"})
```

# Retrieving Memories

```python
store.get(namespace,"1")
```

```python
store.get(namespace2,"2")
```

# Retrieve ALL

```python
items = store.search(namespace)

print(items)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace2)

print(items)

for item in items:
    print(item.value)
```

# Semantic Search

```python
from langgraph.store.memory import InMemoryStore
from langchain_openai import OpenAIEmbeddings
```

```python
embedding_model = OpenAIEmbeddings()
```

```python
store = InMemoryStore(index={'embed': embedding_model, 'dims':1536})
```

```python
namespace = ('users', 'u1')
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(namespace , query="What is user currenlty learning ?" , limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace , query="What are User preferences ?" , limit=3)

for item in items:
    print(item.value)
```
