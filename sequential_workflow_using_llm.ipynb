```python
from langgraph.graph import StateGraph, START,END
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
```

```python
class LLMState(TypedDict):
    question : str
    answer : str
```

```python
llm = ChatOpenAI()
```

```python
def llm_question(LLMState) -> LLMState:
    question = LLMState['question']

    prompt = f"Asnwer this question in 50 words -> {question}"

    response = llm.invoke(prompt)

    LLMState['answer'] = response.content

    return LLMState
```

```python
# create a graph
graph = StateGraph(LLMState)
```

```python
# add node
graph.add_node('question_llm' , llm_question)
#  add edge 
graph.add_edge(START,'question_llm')
graph.add_edge('question_llm' , END)
#  compile graph
workflow = graph.compile()
```

```python
# execute the graph
initial_state = {'question':'How to save yourself from a fire incident'}

final_state = workflow.invoke(initial_state)

print(final_state)
```
